```python
df['order_size'] = df['total_price'].apply(lambda x: 'Large' if x > 100 else 'Small')
df.head()
```

```python
customer_summary = df.groupby('customer_id')['total_price'].agg(
    total_spend='sum',
    orders='count'
).reset_index()

customer_summary['vip_status'] = customer_summary['total_spend'].apply(
    lambda x: 'VIP' if x > 200 else 'Regular'
)

customer_summary.sort_values('total_spend', ascending=False)
```

```python
customers = pd.DataFrame({
    'customer_id': [1001, 1002, 1003, 1004, 1005, 1006, 1007],
    'first_name': ['Jüri', 'Kati', 'Maris', 'Peeter', 'Liina', 'Andres', 'Tiina']
})

merged = pd.merge(customer_summary, customers, on='customer_id', how='left')
merged.sort_values('total_spend', ascending=False)
```
